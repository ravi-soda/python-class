# Lesson 10: K-Means Clustering

Our first **unsupervised** algorithm that *groups* data. There are NO labels — the algorithm discovers groups (clusters) on its own.

Example: a shop wants to group customers into segments ('budget', 'premium', ...) without anyone pre-labeling them.

## The idea

**K-Means** splits data into **K** groups, where K is a number YOU choose. Each group has a **centroid** (its center point). The algorithm:

1. Randomly place K centroids.
2. **Assign** each point to its nearest centroid.
3. **Move** each centroid to the average position of the points assigned to it.
4. Repeat steps 2-3 until centroids stop moving.

It's like dropping K magnets into a cloud of points: each magnet grabs its nearest points, then recenters itself, over and over, until stable.

## Step 1: Make some unlabeled data with natural groups

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# 4 natural blobs, but we PRETEND we don't know the groups
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.8, random_state=42)

plt.scatter(X[:, 0], X[:, 1], s=20, color='gray')
plt.title('Raw data (no labels) — can you see groups?')
plt.grid(True)
plt.show()
```

## Step 2: Run K-Means (K=4)

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)   # assigns each point a cluster number

plt.scatter(X[:, 0], X[:, 1], c=labels, s=20, cmap='viridis')
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
            c='red', marker='X', s=200, label='centroids')
plt.title('K-Means found 4 clusters')
plt.legend()
plt.grid(True)
plt.show()
```

The algorithm grouped the points and found each cluster's center (red X) — using NO labels. That's unsupervised learning.

## Step 3: How do we choose K? The 'Elbow Method'

We try different K values and measure **inertia** (total distance from points to their centroids — lower = tighter clusters). Plot it: the 'elbow' where the curve bends is usually the best K.

```python
inertias = []
K_range = range(1, 10)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    inertias.append(km.inertia_)

plt.plot(K_range, inertias, marker='o')
plt.xlabel('K (number of clusters)')
plt.ylabel('Inertia (lower = tighter)')
plt.title('Elbow Method: the bend suggests the best K')
plt.grid(True)
plt.show()

print('Look for the "elbow" — here it should bend sharply around K=4.')
```

## Key facts

- **Unsupervised**: no labels used. It outputs a cluster number for each point.
- **You choose K** (the number of clusters) — it's a hyperparameter. The elbow method helps.
- **Scale your features first** (like PCA) — K-Means uses distances.
- Limitation: assumes roughly round, similar-sized clusters. (DBSCAN, coming later, handles odd shapes.)
