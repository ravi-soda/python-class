# Lesson 4: Loss Functions & Gradient Descent

**Big idea:** When there's no neat formula to solve for `w` and `b`, we let the computer *walk downhill* into the error valley, one small step at a time, until it reaches the bottom.

We'll use the same data as Lesson 3 and watch gradient descent converge to the same answer (**w ≈ 7.04, b ≈ 27.71**).

> Upload to [Google Colab](https://colab.research.google.com) and run top to bottom.

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.array([1, 2, 3, 4, 5, 6, 7, 8], dtype=float)
y = np.array([35, 42, 50, 55, 62, 68, 78, 85], dtype=float)
n = len(X)
```

## Step 1: The loss function (MSE)

Loss = how wrong the model currently is. We use **Mean Squared Error**:

$$ \text{MSE} = \frac{1}{n}\sum_{i=1}^{n} \big(y_i - (w x_i + b)\big)^2 $$

Lower loss = better fit. Gradient descent's whole job is to make this number small.

```python
def mse(w, b):
    predictions = w * X + b
    return np.mean((y - predictions) ** 2)

# A bad guess vs a good guess
print('Loss at w=0,  b=0    :', round(mse(0, 0), 2))
print('Loss at w=7.04,b=27.71:', round(mse(7.04, 27.71), 2))
```

## Step 2: The gradients (which way is downhill?)

A **gradient** is the slope of the loss with respect to each parameter. It tells us which direction increases the loss — so we step the *opposite* way.

From calculus (derivative of MSE):

$$ \frac{\partial \text{MSE}}{\partial w} = \frac{-2}{n}\sum x_i\,(y_i - (w x_i + b)) $$

$$ \frac{\partial \text{MSE}}{\partial b} = \frac{-2}{n}\sum (y_i - (w x_i + b)) $$

```python
def gradients(w, b):
    error = y - (w * X + b)          # how far off each prediction is
    dw = (-2 / n) * np.sum(X * error)
    db = (-2 / n) * np.sum(error)
    return dw, db
```

## Step 3: The gradient descent loop

Repeat many times:
1. Compute gradients (which way is uphill).
2. Step the *opposite* direction by a small amount = **learning rate**.

$$ w \leftarrow w - \text{lr} \cdot \frac{\partial \text{MSE}}{\partial w} \qquad b \leftarrow b - \text{lr} \cdot \frac{\partial \text{MSE}}{\partial b} $$

```python
w, b = 0.0, 0.0          # start from a bad guess
lr = 0.01                # learning rate (step size)
epochs = 2000            # how many steps

history = []
for step in range(epochs):
    dw, db = gradients(w, b)
    w -= lr * dw
    b -= lr * db
    history.append(mse(w, b))
    if step % 200 == 0:
        print(f'step {step:4d} | w={w:6.3f} | b={b:6.3f} | loss={mse(w,b):8.3f}')

print()
print(f'FINAL: w={w:.3f}, b={b:.3f}  (target from Lesson 3: w=7.04, b=27.71)')
```

## Step 4: Watch the loss shrink

This curve is the model 'walking downhill'. It drops fast, then flattens as it nears the bottom of the valley.

```python
plt.plot(history)
plt.xlabel('Step (epoch)')
plt.ylabel('Loss (MSE)')
plt.title('Gradient descent: loss decreasing over time')
plt.grid(True)
plt.show()
```

## Step 5: Experiment with the learning rate (IMPORTANT)

The learning rate is the single most important knob. Try these and watch what happens:

- `lr = 0.001` → **too small**: learns correctly but very slowly (needs many more epochs).
- `lr = 0.01`  → a good middle ground.
- `lr = 0.05`  → faster, but watch it start to wobble.
- `lr = 0.1`   → **too big**: the loss may explode to `inf`/`nan` (it overshoots the valley and bounces *up* the walls).

Re-run the cell below with different `lr` values.

```python
def train(lr, epochs=2000):
    w, b = 0.0, 0.0
    losses = []
    for _ in range(epochs):
        dw, db = gradients(w, b)
        w -= lr * dw
        b -= lr * db
        losses.append(mse(w, b))
    return w, b, losses

for lr_try in [0.001, 0.01, 0.05]:
    w_t, b_t, losses = train(lr_try)
    plt.plot(losses, label=f'lr={lr_try} (final loss={losses[-1]:.2f})')

plt.xlabel('Step')
plt.ylabel('Loss (MSE)')
plt.title('Effect of learning rate')
plt.legend()
plt.ylim(0, 500)
plt.grid(True)
plt.show()
```
